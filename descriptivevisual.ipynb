```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('museums (1).csv')
df.head()
```

```python
df["Income_Deprivation_Bucket"] = (pd.cut(df["Area_Deprivation_index_income"],bins=[0, 5, 10],labels=["Low deprivation", "High deprivation"]))

counts = pd.crosstab(df["Area_Geodemographic_group"],df["Income_Deprivation_Bucket"])


counts.plot(kind="barh",stacked=False,figsize=(10, 8),color=["red", "Pink"])

plt.title("Number of Museums by City Type and Income Deprivation")
plt.xlabel("Number of Museums")
plt.ylabel("City Type")
plt.legend(title="Income Deprivation")
plt.tight_layout()
plt.show()
```

```python
df["Income_Deprivation_Bucket"] = (pd.cut(df["Area_Deprivation_index_income"],bins=[0, 5, 10],labels=["Low deprivation", "High deprivation"]))

counts = pd.crosstab(df["Governance"],df["Income_Deprivation_Bucket"])


counts.plot(kind="barh",stacked=False,figsize=(14, 9),color=["red", "Pink"])

plt.title("Number of Museums by Governance and Income Deprivation")
plt.xlabel("Number of Museums")
plt.ylabel("Governance")
plt.legend(title="Income Deprivation")
plt.show()
```

```python
df["Income_Deprivation_Bucket"] = (pd.cut(df["Area_Deprivation_index_income"],bins=[0, 5, 10],labels=["Low deprivation", "High deprivation"]))
df["Health_Bucket"] = (pd.cut(df["Area_Deprivation_index_health"],bins=[0, 5, 10],labels=["Low Health", "High Health"]))


counts = pd.crosstab(df["Health_Bucket"],df["Income_Deprivation_Bucket"])


counts.plot(kind="barh",stacked=False,figsize=(14, 9),color=["red", "Pink"])

plt.title("Number of Museums by Size and Income Deprivation")
plt.xlabel("Number of Museums")
plt.ylabel("average health within area")
plt.legend(title="Income Deprivation")
plt.show()
```
